```python
import requests
```

```python
def fileHandling():
    try:
        with open("api.txt", "r") as f:
            for line in f:
                line = line.strip()
            if line:
                return line
            print("The api file is empty.")
            return None
    except FileNotFoundError:
        print("The file does not exist.")
        return None
```

```python
def main():
    URL = fileHandling()

    if not URL:
        raise Exception("No URL found, exiting with error code 1.")
        return 
    
    print("Fetching data from API....")

    # exception handling for internet connection
    try:
        response = requests.get(URL)
        print("Data successfully fetched.")
    except requests.exceptions.RequestException:
        print("Connection Failed. Check your Internet Connection.")
        return
    
    # condition for HTTP status code
    if response.status_code != 200:
        print(f"HTTP Error: {response.status_code}")
        return
    
    data = response.json()

    # API meta data printing
    print("\n \n")
    print("\t \tAPI META DATA\t \t")
    for key, value in data.items():
        print(f"{key}: {value}")
    print("_" * 30)
    

    # printing match data
    print("\n \n")
    print("\t \tCricket Data\t \t")
    for matchData in data['data']:
        print('_' * 30)
        print(f"Match: {matchData.get("name")}")
        print(f"Status: {matchData.get("status")}")
        print(f"Venue: {matchData.get("venue")}")
        print(f"Date: {matchData.get("date")}")

        if 'score' in matchData:
            print("." * 30)
            for scores in matchData['score']:
                print(f"Innings: {scores.get('innings')}")
                print(f"Runs: {scores.get('r')}")
                print(f"Wickets: {scores.get('w')}")
                print(f"Overs: {scores.get('o')}")
        else:
            print("No score data for this match.")
    
    print("_" * 30)
    
```

```python
main()
```
